# 5. Logic: Conditional Statements

In coding, it can be useful to execute some code if a condition is true and do something else if not. This allows our code to have flexibility and handle different use cases.

Python handles conditions with the use of booleans: `True` and `False`.

Booleans are technically also integers. `True` is 1 while `False` is 0. This can be very useful with lists and the `sum` function, where you can check how many elements in a list are `True`.

```python
test_list = [True, False, True, True, False]
sum(test_list)
```

There are a number of common "mathematical" operations with booleans that allow one to do more complex things. The most common are `and` and `or`. In Python, the use of these is very straightforward:

```python
print(True and False) # the `and` operator only returns True if both the left and right side are true
print(True or False) # the `or` operator returns True if either the left and right side are true
```

It is also possible to check equivalence (`==`, `!=`) and relative size (`>`, `<`, `<=`, etc) in Python:

```python
print(3 == 3) # check if two objects are the same
print([4,5,2] == [4,5,2]) # works for all python objects
print(4 != 5) # returns True if two things are not the same
print(3 > 8) # the same as its mathematical equivalent
```

`if/else` is used to make your code do certain things when something is true. This needs to be done using `if` and a colon with indentation. The syntax is:

```python
'''
if [boolean expression]:
  do something
else:
  do something else
'''
```

The boolean expression must be able to be evaluated as `True` or `False`. The expression under `else` is evaluated if the statement next to `if` is `False`. Here's a simple example:

```python
num = 1
if num > 2:
  print('hello')
else:
  print('whoops')

# What will the code print if we change 3 to 1 above?
#prints whoops
```

If you want to use multiple `if` statements, you can use `elif`, which is short for `else if`

```python
num = 3
if num > 2: # evaluated first
  print('This number is greater than 2')
elif num > 0: # evaluated if the statement above is False
  print('This number is greater than 0')
else: # evaluated if all if/elif statements are False
  print('This number is less than or equal to 0')

# What will the code print if we change 3 to 1 above? to -1?
# for 1, prints that its greater than 0. for -1, frints that its less than or equal to -1
```
